```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Everything is working!")
```

```python
data = pd.read_csv("Car Dekho Data.csv")
data.head()
```

```python
print("Shape of dataset:", data.shape)
print("\nColumns:")
print(data.columns.tolist())

print("\nDataset information:")
data.info()
```

```python
data.isnull().sum()
```

```python
data.describe()
```

```python
print("Fuel Types:")
print(data["Fuel_Type"].value_counts())

print("\nSeller Types:")
print(data["Seller_Type"].value_counts())

print("\nTransmission Types:")
print(data["Transmission"].value_counts())

print("\nOwner:")
print(data["Owner"].value_counts())
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x="Fuel_Type", y="Selling_Price", data=data)
plt.title("Selling Price by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Selling Price")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x="Present_Price", y="Selling_Price", data=data)
plt.title("Selling Price vs Present Price")
plt.xlabel("Present Price")
plt.ylabel("Selling Price")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x="Kms_Driven", y="Selling_Price", data=data)
plt.title("Selling Price vs Kilometers Driven")
plt.xlabel("Kilometers Driven")
plt.ylabel("Selling Price")
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(x="Transmission", y="Selling_Price", data=data)
plt.title("Selling Price by Transmission Type")
plt.xlabel("Transmission")
plt.ylabel("Selling Price")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(x="Year", y="Selling_Price", data=data)
plt.title("Selling Price vs Manufacturing Year")
plt.xlabel("Year")
plt.ylabel("Selling Price")
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(x="Seller_Type", y="Selling_Price", data=data)
plt.title("Selling Price by Seller Type")
plt.xlabel("Seller Type")
plt.ylabel("Selling Price")
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(x="Owner", y="Selling_Price", data=data)
plt.title("Selling Price by Number of Previous Owners")
plt.xlabel("Number of Previous Owners")
plt.ylabel("Selling Price")
plt.show()
```

```python
data[["Year", "Selling_Price", "Present_Price", "Kms_Driven", "Owner"]].corr()
```

```python
plt.figure(figsize=(9, 6))

corr = data[["Year", "Selling_Price", "Present_Price", "Kms_Driven", "Owner"]].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap of Car Market Variables")
plt.show()
```

```python
data.groupby("Transmission")["Selling_Price"].mean().sort_values(ascending=False)
```
